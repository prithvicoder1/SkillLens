```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('titanic.csv')
```

```python
df.head()
```

```python
df['number'].unique()
```

```python
fig = df['number'].value_counts().plot.bar()
fig.set_title('Passengers travelling with')
```

```python
# Extract numerical part
df['number_numerical'] = pd.to_numeric(df["number"],errors='coerce',downcast='integer')
# error='coerce' - means that Pandas converts invalid values to NaN.
#downcast - means that After conversion, Pandas tries to use the smallest possible integer type to save memory.
```

```python
# Extra Categorical Part
df['number_categorical'] = np.where(df['number_numerical'].isnull(),df['number'],np.nan)
df.head()
```

```python
df['Cabin'].unique
```

```python
df['cabin_num'] = pd.to_numeric(
    df['Cabin'].str.extract(r'(\d+)')[0],
    errors='coerce'
)
```
